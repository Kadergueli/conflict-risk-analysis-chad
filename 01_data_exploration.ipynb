```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert LATITUDE and LONGITUDE to numeric, handling commas as decimal separators
df['LATITUDE'] = df['LATITUDE'].str.replace(',', '.').astype(float)
df['LONGITUDE'] = df['LONGITUDE'].str.replace(',', '.').astype(float)

# Display the dtypes to confirm the conversion
print(df[['LATITUDE', 'LONGITUDE']].dtypes)
```

```python
# boxplot of FATALITIES by EVENT_TYPE
plt.figure(figsize=(12, 8))
sns.boxplot(x='EVENT_TYPE', y='FATALITIES', data=df)
plt.title('Distribution of Fatalities by Event Type')
plt.xlabel('Event Type')
plt.ylabel('Fatalities')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
